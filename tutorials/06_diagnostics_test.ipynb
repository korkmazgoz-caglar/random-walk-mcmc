```python
import numpy as np
from rwmcmc import random_walk_metropolis_hastings, banana_log_pdf, dashboard

rng = np.random.default_rng(42)
samples, accepted = random_walk_metropolis_hastings(
    banana_log_pdf, x0=[0.0, 3.0], n_samples=100000, step_size=[4.0, 2.0], rng=rng
)
fig = dashboard(samples, accepted, burn_in=1000, param_names=["x1", "x2"])
```
